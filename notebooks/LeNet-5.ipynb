```python
import tensorflow as tf
import numpy as np

SEED = 42
tf.keras.utils.set_random_seed(SEED)

print("=" * 70)
print("BUILDING LENET-5")
print("=" * 70)

lenet5 = tf.keras.Sequential([

    tf.keras.layers.Input(shape=(32, 32, 3), name="input"),

    # C1
    tf.keras.layers.Conv2D(
        6,
        kernel_size=(5, 5),
        activation="tanh",
        padding="valid",
        name="conv1"
    ),

    # S2
    tf.keras.layers.AveragePooling2D(
        pool_size=(2, 2),
        strides=2,
        name="pool1"
    ),

    # C3
    tf.keras.layers.Conv2D(
        16,
        kernel_size=(5, 5),
        activation="tanh",
        padding="valid",
        name="conv2"
    ),

    # S4
    tf.keras.layers.AveragePooling2D(
        pool_size=(2, 2),
        strides=2,
        name="pool2"
    ),

    # C5
    tf.keras.layers.Conv2D(
        120,
        kernel_size=(5, 5),
        activation="tanh",
        padding="valid",
        name="conv3"
    ),

    tf.keras.layers.Flatten(name="flatten"),

    # F6
    tf.keras.layers.Dense(
        84,
        activation="tanh",
        name="dense1"
    ),

    # Output — 28 traffic-sign classes
    tf.keras.layers.Dense(
        28,
        activation="softmax",
        name="predictions"
    )
], name="LeNet5_TSR")

# Compile
lenet5.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=0.001),
    loss="categorical_crossentropy",
    metrics=["accuracy"]
)

# Summary
lenet5.summary()

print("\n" + "=" * 70)
print("INPUT SHAPE :", lenet5.input_shape)
print("OUTPUT SHAPE:", lenet5.output_shape)
print("PARAMETERS  :", lenet5.count_params())
print("=" * 70)
```

```python
import tensorflow as tf
import numpy as np

def make_lenet_gradcam(image_batch, model, last_conv_layer_name):
    """
    Generate Grad-CAM heatmap for LeNet-5.

    Parameters:
        image_batch: Tensor of shape (1, 32, 32, 3)
        model: trained LeNet-5 model
        last_conv_layer_name: name of the final Conv2D layer

    Returns:
        heatmap: normalized Grad-CAM heatmap
        predicted_class: predicted class index
    """

    # Model that returns:
    # 1. convolutional feature maps
    # 2. final predictions
    grad_model = tf.keras.models.Model(
        inputs=model.inputs,
        outputs=[
            model.get_layer(last_conv_layer_name).output,
            model.output
        ]
    )

    with tf.GradientTape() as tape:

        # Forward pass
        conv_outputs, predictions = grad_model(
            image_batch,
            training=False
        )

        # Predicted class
        predicted_class = tf.argmax(
            predictions[0]
        )

        # Score of predicted class
        class_score = predictions[:, predicted_class]

    # Gradient of class score with respect to feature maps
    grads = tape.gradient(
        class_score,
        conv_outputs
    )

    # Global average pooling of gradients
    pooled_grads = tf.reduce_mean(
        grads,
        axis=(1, 2)
    )

    # Remove batch dimension
    conv_outputs = conv_outputs[0]
    pooled_grads = pooled_grads[0]

    # Weighted combination of feature maps
    heatmap = tf.reduce_sum(
        conv_outputs * pooled_grads,
        axis=-1
    )

    # ReLU
    heatmap = tf.maximum(
        heatmap,
        0
    )

    # Normalize to [0, 1]
    max_value = tf.reduce_max(heatmap)

    heatmap = tf.where(
        max_value > 0,
        heatmap / max_value,
        tf.zeros_like(heatmap)
    )

    # Resize heatmap to input image size (32x32)
    heatmap = tf.image.resize(
        heatmap[..., tf.newaxis],
        (32, 32)
    )

    heatmap = tf.squeeze(
        heatmap,
        axis=-1
    )

    return (
        heatmap.numpy(),
        int(predicted_class.numpy())
    )


print("=" * 70)
print("LENET-5 GRAD-CAM FUNCTION CREATED")
print("=" * 70)

# Show Conv2D layers so we can select the correct target layer
print("\nLeNet-5 Conv2D layers:")

for layer in lenet5.layers:
    if isinstance(layer, tf.keras.layers.Conv2D):
        print(
            f"Layer name: {layer.name} | "
            f"Output shape: {layer.output.shape}"
        )
```

```python
import numpy as np
import pandas as pd
import tensorflow as tf

all_heatmaps = []
all_labels = []
all_predictions = []
all_confidences = []

print("=" * 70)
print("LENET-5 FULL TEST-SET ATTENTION ANALYSIS")
print("=" * 70)

for images, labels in lenet_test_norm:

   
    with tf.GradientTape() as tape:

        x = gradcam_lenet.get_layer("conv1")(
            images, training=False
        )

        x = gradcam_lenet.get_layer("pool1")(
            x, training=False
        )

        conv2_output = gradcam_lenet.get_layer("conv2")(
            x, training=False
        )

        tape.watch(conv2_output)

        x = gradcam_lenet.get_layer("pool2")(
            conv2_output, training=False
        )

        x = gradcam_lenet.get_layer("conv3")(
            x, training=False
        )

        x = gradcam_lenet.get_layer("flatten")(
            x, training=False
        )

        x = gradcam_lenet.get_layer("fc1")(
            x, training=False
        )

        classifier = gradcam_lenet.get_layer("predictions")

        logits = (
            tf.linalg.matmul(
                x,
                classifier.kernel
            )
            + classifier.bias
        )

        probabilities = tf.nn.softmax(logits)

        predicted_classes = tf.argmax(
            logits,
            axis=1
        )

        # Select each sample's predicted-class logit
        batch_indices = tf.range(
            tf.shape(logits)[0],
            dtype=tf.int64
        )

        gather_indices = tf.stack(
            [
                batch_indices,
                tf.cast(
                    predicted_classes,
                    tf.int64
                )
            ],
            axis=1
        )

        selected_logits = tf.gather_nd(
            logits,
            gather_indices
        )

        # Sum keeps all samples connected to the graph
        total_selected_logit = tf.reduce_sum(
            selected_logits
        )

    

    grads = tape.gradient(
        total_selected_logit,
        conv2_output
    )

    if grads is None:
        raise RuntimeError(
            "Batch-wise gradient computation failed."
        )


    predicted_classes = predicted_classes.numpy()

    true_classes = tf.argmax(
        labels,
        axis=1
    ).numpy()

    confidences = tf.reduce_max(
        probabilities,
        axis=1
    ).numpy()

    conv2_np = conv2_output.numpy()
    grads_np = grads.numpy()

  

    for i in range(images.shape[0]):

        feature_map = conv2_np[i]
        gradient_map = grads_np[i]

        # Global average pooling of gradients
        weights = np.mean(
            gradient_map,
            axis=(0, 1)
        )

        # Signed response
        raw_map = np.sum(
            feature_map * weights,
            axis=-1
        )

       
        attention_map = np.abs(
            raw_map
        )

        # Normalize to [0,1]
        min_val = attention_map.min()
        max_val = attention_map.max()

        if max_val > min_val:

            attention_map = (
                attention_map - min_val
            ) / (
                max_val - min_val
            )

        else:

            attention_map = np.zeros_like(
                attention_map
            )


        attention_map = tf.image.resize(
            attention_map[..., np.newaxis],
            (32, 32),
            method="bilinear"
        ).numpy().squeeze()

        # Store
        all_heatmaps.append(
            attention_map.flatten()
        )

        all_labels.append(
            true_classes[i]
        )

        all_predictions.append(
            predicted_classes[i]
        )

        all_confidences.append(
            confidences[i]
        )


all_heatmaps = np.array(
    all_heatmaps
)

all_labels = np.array(
    all_labels
)

all_predictions = np.array(
    all_predictions
)

all_confidences = np.array(
    all_confidences
)

print("\nProcessed images:", len(all_heatmaps))


important_pixel_percentages = (
    np.mean(
        all_heatmaps >= 0.5,
        axis=1
    ) * 100
)

average_heatmap_intensities = (
    np.mean(
        all_heatmaps,
        axis=1
    )
)

prediction_accuracy = (
    np.mean(
        all_labels == all_predictions
    ) * 100
)


normalized_heatmaps = (
    all_heatmaps /
    (
        np.linalg.norm(
            all_heatmaps,
            axis=1,
            keepdims=True
        ) + 1e-12
    )
)


overall_centroid = np.mean(
    normalized_heatmaps,
    axis=0
)

overall_centroid = (
    overall_centroid /
    (
        np.linalg.norm(
            overall_centroid
        ) + 1e-12
    )
)

overall_attention_consistency = np.mean(
    normalized_heatmaps @
    overall_centroid
)

classwise_results = []

for class_id in range(28):

    indices = np.where(
        all_labels == class_id
    )[0]

    if len(indices) == 0:
        continue

    class_maps = normalized_heatmaps[
        indices
    ]

    class_centroid = np.mean(
        class_maps,
        axis=0
    )

    class_centroid = (
        class_centroid /
        (
            np.linalg.norm(
                class_centroid
            ) + 1e-12
        )
    )

    similarities = (
        class_maps @
        class_centroid
    )

    classwise_results.append({
        "Class": class_id,
        "Images": len(indices),
        "Attention_Consistency":
            np.mean(similarities)
    })

classwise_df = pd.DataFrame(
    classwise_results
)


summary_df = pd.DataFrame([{
    "Test Images":
        len(all_heatmaps),

    "Important Pixels Percentage (%)":
        np.mean(
            important_pixel_percentages
        ),

    "Average Heatmap Intensity":
        np.mean(
            average_heatmap_intensities
        ),

    "Overall Attention Consistency":
        overall_attention_consistency,

    "Prediction Accuracy (%)":
        prediction_accuracy
}])


print("\n" + "=" * 70)
print("LENET-5 QUANTITATIVE ATTENTION RESULTS")
print("=" * 70)

print(
    f"Test Images                   : "
    f"{len(all_heatmaps)}"
)

print(
    f"Mean Important Pixels (%)     : "
    f"{np.mean(important_pixel_percentages):.4f}%"
)

print(
    f"Mean Heatmap Intensity        : "
    f"{np.mean(average_heatmap_intensities):.4f}"
)

print(
    f"Overall Attention Consistency : "
    f"{overall_attention_consistency:.4f}"
)

print(
    f"Prediction Accuracy           : "
    f"{prediction_accuracy:.2f}%"
)

print("=" * 70)

print("\nClass-wise attention consistency:")
print(
    classwise_df.to_string(
        index=False
    )
)


summary_df.to_csv(
    "LeNet5_GradCAM_Final_Summary.csv",
    index=False
)

classwise_df.to_csv(
    "LeNet5_Classwise_Attention_Consistency.csv",
    index=False
)

sample_results_df = pd.DataFrame({
    "True_Class":
        all_labels,

    "Predicted_Class":
        all_predictions,

    "Confidence":
        all_confidences,

    "Important_Pixels_Percentage":
        important_pixel_percentages,

    "Average_Heatmap_Intensity":
        average_heatmap_intensities
})

sample_results_df.to_csv(
    "LeNet5_GradCAM_Sample_Results.csv",
    index=False
)

print("\nFiles saved successfully.")
```
